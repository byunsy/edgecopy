```python
import os
import sys
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import metric as mt
import matplotlib.ticker as ticker
from glob import glob
```

```python
del_data = [
    [0.01, "2-copy", 0.999150],
    [0.05, "2-copy", 0.999816],
    [0.10, "2-copy", 0.999704],
    [0.15, "2-copy", 0.999690],
    [0.20, "2-copy", 0.999590],
    
    [0.01, "3-copy", 0.996992],
    [0.05, "3-copy", 0.996942],
    [0.10, "3-copy", 0.996743],
    [0.15, "3-copy", 0.996348],
    [0.20, "3-copy", 0.994994],
]

df_del = pd.DataFrame(del_data)
df_del.columns = ['prop','Gene Type','conc']

dup_data = [
    [0.01, "2-copy", 0.998423],
    [0.05, "2-copy", 0.998442],
    [0.10, "2-copy", 0.997891],
    [0.15, "2-copy", 0.997393],
    [0.20, "2-copy", 0.996987],
    
    [0.01, "3-copy", 0.996608],
    [0.05, "3-copy", 0.996460],
    [0.10, "3-copy", 0.995158],
    [0.15, "3-copy", 0.992618],
    [0.20, "3-copy", 0.990116],
]

df_dup = pd.DataFrame(dup_data)
df_dup.columns = ['prop','Gene Type','conc']
```

```python
plt.figure(figsize=(8, 6))


ax = sns.lineplot(df_del, x='prop', y='conc', hue='Gene Type',
                  marker='o', markersize=8, palette='Set1')

# Formatting
ax.grid(which='major', color='#BBBBBB', linewidth=0.8)
ax.grid(which='minor', color='#BBBBBB', linestyle=':', linewidth=0.5)
ax.minorticks_on()
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.3f'))

plt.legend(title='Gene Type', loc='lower left')
plt.title('Edgecopy on Simulated CNVs (-1 Deletion)', fontsize=14)
plt.xlabel('CNV frequency', fontsize=14)
plt.ylabel('Concordance', fontsize=14)
plt.ylim(0.98, 1.003)
plt.show()
```

```python
plt.figure(figsize=(8, 6))


ax = sns.lineplot(df_dup, x='prop', y='conc', hue='Gene Type',
                  marker='o', markersize=8, palette='Set1')

# Formatting
ax.grid(which='major', color='#BBBBBB', linewidth=0.8)
ax.grid(which='minor', color='#BBBBBB', linestyle=':', linewidth=0.5)
ax.minorticks_on()
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.3f'))

plt.legend(title='Gene Type', loc='lower left')
plt.title('Edgecopy on Simulated CNVs (+1 Duplication)', fontsize=14)
plt.xlabel('CNV frequency', fontsize=14)
plt.ylabel('Concordance', fontsize=14)
plt.ylim(0.98, 1.003)
plt.show()
```
